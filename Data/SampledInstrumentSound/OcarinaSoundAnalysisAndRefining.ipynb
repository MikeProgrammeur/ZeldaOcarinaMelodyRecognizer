In this file we are going to open the ocarina sample, downloaded at https://freewavesamples.com/kawai-k5000w-ocarina-c5. We will remove the silence at the beginning and at the end. We will also identify the end of the attack and the beginning of the realease. This will allow us to remove the attack and release between two notes to have a smooth transition instead of juxtaposed notes.

```python
import numpy as np
import scipy.signal as sig
from scipy.io import wavfile


import IPython.display as ipd
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
OcarinaC5Rate, OcarinaC5Data = wavfile.read('./Kawai-K5000W-Ocarina-C5.wav')
print(f"The rate is {OcarinaC5Rate}")
print(f"The shape of the data matrix is {OcarinaC5Data.shape}")
```

```python
OcarinaC5Data = OcarinaC5Data[:,0] # Keep only one channel
OcarinaC5Data = OcarinaC5Data/np.max(np.abs(OcarinaC5Data)) # Normalize
print(f"Now the shape of the data matrix is {OcarinaC5Data.shape} and its extremum is {np.max(np.abs(OcarinaC5Data))}")
```

### Let us hear it

```python
ipd.Audio(OcarinaC5Data, rate=OcarinaC5Rate)
```

There are silences at the begining and at the end

### Let us see it

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    y=OcarinaC5Data,
    mode='lines',
    name='Signal',
    line=dict(color='blue')
))

fig.update_layout(
    title='Signal Through Time',
    xaxis_title='Time ',
    yaxis_title='Magnitude [no unit]',
)

fig.show()
```

```python
ipd.Audio(OcarinaC5Data[1500:45111], rate=OcarinaC5Rate)
```

We can the end of the release is around 45200 (in term of  array index). There is still signal after but it is weak and must some echoes.

```python
OcarinaC5DataCut = OcarinaC5Data[:45200]
```

```python
ipd.Audio(OcarinaC5DataCut, rate=OcarinaC5Rate)
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    y=OcarinaC5DataCut,
    mode='lines',
    name='Signal',
    line=dict(color='blue')
))

fig.update_layout(
    title='Signal Through Time',
    xaxis_title='Time ',
    yaxis_title='Magnitude [no unit]',
)

fig.show()
```

```python
amplitude = np.iinfo(np.int16).max

data = amplitude * OcarinaC5DataCut

wavfile.write("OcarinaK5000W_C5Refined.wav", OcarinaC5Rate, data.astype(np.int16))
```

And now we can roughly identify the attack and release.\
We will consider the attack to end at 800 and the release to start at 44600.\
We will save those number in a file next to the .wav, I think it is a better practice to store it in a file rather than hard code it in another python script.


```python
with open('./OcarinaK5000W_C5RefinedAttackRelease.npy', 'wb') as f:
    np.save(f, np.array([800, 44600]))
```
