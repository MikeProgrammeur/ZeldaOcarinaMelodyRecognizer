### Generate song from synthetic sinus curves

```python
def synthetise_song(list_of_freq : List[int], fe : float, note_duration : float = 1, list_of_harmonics : List[float] = [1]):
    note_duration_stamps = int(note_duration * fe)
    signal_length = note_duration_stamps * len(list_of_freq)
    t_array = np.arange(signal_length)
    f_array = np.repeat(np.array(list_of_freq), note_duration_stamps)
    sig = np.zeros(signal_length)
    for i,harmonic in enumerate(list_of_harmonics):
        sig += harmonic * np.sin(2*np.pi*i*f_array*t_array/fe)
    return sig
```

```python
def synthetise_song(
    list_of_freq: List[float],
    fe: float,
    note_duration: List[float],
    list_of_harmonics: List[float] = [1.0],
) -> np.ndarray:
    if len(list_of_freq) != len(note_duration):
        raise ValueError("list_of_freq and note_duration must have the same length.")

    note_signals = []

    for freq, duration in zip(list_of_freq, note_duration):
        n_samples = int(duration * fe)
        t = np.arange(n_samples) / fe
        sig = np.zeros(n_samples)

        for k, amplitude in enumerate(list_of_harmonics, start=1):
            sig += amplitude * np.sin(2 * np.pi * k * freq * t)

        note_signals.append(sig)

    return np.concatenate(note_signals) if note_signals else np.array([])
```

### Generate song from one Ocarina 440Hz tone

```python
# load the sample at C5 = 523.25 Hz
refinedRate, refinedData = wavfile.read('./Data/SampledInstrumentSound/OcarinaK5000W_C5Refined.wav')

with open('./Data/SampledInstrumentSound/OcarinaK5000W_C5RefinedAttackRelease.npy', 'rb') as f:
    attackReleaseIdxOcarina = np.load(f) 
print(attackReleaseIdxOcarina)

ocarina_pitch = 523.25 # Hardcoding is bad, TODO write this in the .npy file
```

```python
def adjustSong( 
    x : np.ndarray,
    fs : int,
    origPitch : float,
    targetPitch : float,
    target_duration : float,
    attackReleaseIdx : np.ndarray,
    withAttack : bool = False,
    withRelease : bool = False
    ) -> np.ndarray:
    
    if attackReleaseIdx[0] < 0 or attackReleaseIdx[1] < attackReleaseIdx[0] or len(x) < attackReleaseIdx[1] :
        raise AssertionError("Please ensure the attack and release indices are consistent with the length of the signal")
    
    if fs <= 0 or origPitch < 0 or targetPitch < 0 or target_duration < 0:
        raise AssertionError(f"One parameter is lower than 0 and should not fs: {fs}, origPitch: {origPitch}, targetPitch: {targetPitch}, target_duration: {target_duration}")
    
    startIdx = 0 if withAttack else attackReleaseIdx[0]
    stopIdx = len(x) if withRelease else attackReleaseIdx[1]
    
    x_cut = x[startIdx:stopIdx]
    
    x_cut_resamp = resample(x_cut, int(fs * target_duration), t=None, axis=0, window=None, domain='time')
    
    # When we do this resampling the pitch is shifted, new_pitch = orig_pitch * old_duration / new_duration
    old_duration = len(x_cut)/fs
    new_pitch = origPitch * old_duration / target_duration
    
    # Convert the Hz shift into the number of steps to shift
    n_step_shift = 12*np.log2(targetPitch/new_pitch)
    
    x_cut_resamp_shifted = librosa.effects.pitch_shift(x_cut_resamp, sr=fs, n_steps=n_step_shift)
    
    return x_cut_resamp_shifted
```

```python
def synthetise_song_from_sample(
    list_of_freq: List[float],
    fs: float,
    note_duration: List[float],
    ocarina_sample: np.ndarray,
    ocarina_sample_pitch: float,
    attackReleaseIdx: np.ndarray,
    list_of_attack: np.ndarray = None, # 1 with attack, 2 with release, 3 with both, 0 with nothing
    )-> np.ndarray:
    note_signals = []
    for i in range(len(list_of_freq)):
        if list_of_attack == None :
            withAttack = False
            withRelease = False
        else:
            withAttack = (list_of_attack[i] == 1 or list_of_attack[i] == 3)
            withRelease = (list_of_attack[i] == 2 or list_of_attack[i] == 3)
            
        if withAttack: # we add a 20ms silence
            note_signals.append(np.zeros(int(0.02*fs)))
            
        note_signals.append(adjustSong(x=ocarina_sample, fs=fs, origPitch=ocarina_sample_pitch, targetPitch = list_of_freq[i], target_duration = note_duration[i], attackReleaseIdx = attackReleaseIdx, withAttack = False, withRelease = False))
        
        if withRelease: # we add a 50ms silence
            note_signals.append(np.zeros(int(0.05*fs)))
    return np.concatenate(note_signals) if note_signals else np.array([])
    
```

# Final step: Convert the songs to MIDI and compute their first-order differences
This makes the recognition independent of the octave in which the user whistles, while also allowing pitch shifts by non-octave intervals.

```python
song_dict_note_midi = {}
song_dict_note_midi_diff = {}

for index, (key, value) in enumerate(song_dict_freq.items()):
    value_midi = 69 + 12*np.log2((np.array(value))/440)

    song_dict_note_midi[key] = value_midi
    song_dict_note_midi_diff[key] = np.diff(value_midi)
    
```
